# Project 004: Decision Tree Classifier for Iris Dataset

This notebook demonstrates an end-to-end multi-class classification pipeline using Decision Trees:
1. **Data Loading & Inspection**: Examining Fisher's Iris dataset.
2. **Stratified Splitting**: Partitioning into train and test sets.
3. **Model Training**: Fitting a `DecisionTreeClassifier`.
4. **Evaluation**: Assessing Accuracy, Confusion Matrix, and Classification Report.
5. **Rule Extraction**: Inspecting the learned tree splits and feature importances.

```python
from main import load_iris_dataframe, train_decision_tree, evaluate_classifier
from sklearn.model_selection import train_test_split
from sklearn.tree import export_text
import pandas as pd

# 1. Load data
df, feature_names, target_names = load_iris_dataframe()
df.head()
```

```python
# 2. Train-test split (stratified)
X = df[feature_names].values
y = df["target"].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# 3. Train Decision Tree model
model = train_decision_tree(X_train, y_train, max_depth=3, criterion="gini")

# 4. Evaluate metrics
metrics = evaluate_classifier(model, X_test, y_test, target_names)
print(f"Test Accuracy: {metrics['accuracy'] * 100:.2f}%")
print(metrics["classification_report"])
```

```python
# 5. Inspect tree rules and feature importances
print("Learned Rules:")
print(export_text(model, feature_names=feature_names))

imp_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
}).sort_values(by="Importance", ascending=False)
imp_df
```

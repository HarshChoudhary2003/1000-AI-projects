# Project 003: K-Means Clustering for Customer Segmentation

This notebook demonstrates an end-to-end unsupervised learning pipeline for retail customer segmentation:
1. **Data Generation & Inspection**: Simulating realistic customer income and spending scores.
2. **Feature Scaling**: Standardizing attributes using `StandardScaler`.
3. **Optimal K Diagnostics**: Evaluating the Elbow Method (WCSS/Inertia) and Silhouette Scores.
4. **K-Means Model Training**: Fitting the model with optimal clusters.
5. **Cluster Profiling & Persona Mapping**: Business interpretation of customer segments.

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# 1. Generate realistic customer data
from main import generate_customer_dataset, evaluate_k_values, profile_segments

df = generate_customer_dataset(n_samples=500, random_state=42)
df.head()
```

```python
# 2. Prepare and scale features
feature_cols = ["Annual_Income_k$", "Spending_Score"]
X = df[feature_cols].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Evaluate optimal K values
k_range, inertias, silhouettes = evaluate_k_values(X_scaled, range(2, 8))

eval_df = pd.DataFrame({
    "K": k_range,
    "Inertia_WCSS": [round(i, 2) for i in inertias],
    "Silhouette_Score": [round(s, 4) for s in silhouettes]
})
eval_df
```

```python
# 4. Fit K-Means with chosen K=5
kmeans = KMeans(n_clusters=5, init='k-means++', n_init=10, random_state=42)
df["Cluster"] = kmeans.fit_predict(X_scaled)

# 5. Profile and review segment personas
summary = profile_segments(df, "Cluster")
summary
```
